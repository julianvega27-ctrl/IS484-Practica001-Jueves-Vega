## EJERCICIO 02: Escalamiento de tickets de soporte técnico

### FICHA PEAS

>**Objetivo:** Resolver cada ticket en el nivel adecuado, atendiendo primero lo tickets críticos.
>
>**Medidad de desempeño (P):** Cumplimiento de los acuerdos de nivel de servicio, tiempo promedio de resolución, satisfacción del cliente y proporción de tickets esperados que el nivel 1 pudo resolver.
>
>**Entorno (E):** Mesa de ayuda a una empresa de servicios, con tickets que llegan continuamente, y niveles con recursos limitados. La urgencia la declara el cliente y puede no tener verdaderamente el nivel que solicita.
>
>**Acciones (A):** Mantener el nivel 1 (mesa de ayuda), escalar al nivel 2 (Soporte especializado) o escalar a nivel 3 (Ingeniería).
>
>**Percepción (S):** tiempo_espera_minutos (entero), nivel_urgencia (alta, baja, media) y cliente premium (True o False).

### REGLAS DE DESICIÓN

|Urgencia|Condición adicional|Acción|
|---|---|---|
|Alta|Cualquier tiempo y tipo de cliente|Escalar a nivel 3|
|Media|cliente_premium o tiempo de espera > 30|Escalar a nivel 2|
|Media|resto|Mantener en nivel 1|
|Baja|cliente_premium y tiempo_espera > 60|Escalar a nivel 2|
|Baja|tiempo_espera > 120|Escalar a nivel 2|
Baja|resto|Mantener en nivel 1|

### JUSTIFICACIÓN DE LAS REGLAS
La urgencia alta se escala a nivel 3 de inmediato sin alguna otra condición, ya que un servicio caído genera pérdidas cada minuto que espera en cola. La urgencia media es condicional ya que basta que alguna de las otras condiciones se presente cómo por ejemplo ser un cliente premium con un SLA comprometido o llevar más de 30 minutos esperando es suficiente para escalarlo, ya que ninguna de las 2 solas satura al equipo especializado. Para escalar en la urgencia baja se necesita de ambas condiciones o un tiempo muy prolongado, para no saturar al segundo nivel y de lo contrario, se debería mantener en el nivel 1.
