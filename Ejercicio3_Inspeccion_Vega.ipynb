## EJERCICIO 03: Inspección Visual del Producto

### FICHA PEAS

>**Objetivo:** Imperdir que productos defectuosos lleguen al cliente, sin descartar productos que esten en buen estado.
>
>**Medidad de desempeño (P):** Proporción de productos defectuosos detectados correctamente, proporción de productos bueno rechazados por error, carga de trabajo enviada a revisión manual y productos inspeccionados.
>
>**Entorno (E):** Producción industrial, donde los productos pasan de forma constante la cámara. Es parcialmente observable, ya que la cámara solo ve una cara del producto y con una baja resolución.
>
>**Acciones (A):** Aprobar el producto, enviarlo a revisión manual o rechazarlo.
>
>**Percepción (S):** Una matriz de 5x5 capturada por cámara, donde el valor es "0" un pixel noraml o "1" un pixel defectuoso.

### REGLAS DE DESICIÓN
|Píxeles defectuoso|Acción|
|---|---|
|0|Aprobar|
|1 a 3|Enviar a revisión manual|
|más de 3|Rechazar|

### CÓDIGO

```python
import numpy as np

def agente_inspeccion (imagen):

    defectos = int(np.sum(imagen == 1))

    if defectos == 0:
        return ("Aprobar", "No se detectaron pixeles defectuosos")
    elif defectos <= 3:
        return ("Revisión manual", 
                 f"Se detectaron {defectos} píxel(es) defectuoso(s): Evidencia ambigua")
    else:
        return ("Rechazar", f"Se detectaron {defectos} pixeles defectuosos")
```
